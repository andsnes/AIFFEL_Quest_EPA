Step 2 : TextSplitters 사용해보기

```python
from langchain_text_splitters import CharacterTextSplitter, RecursiveCharacterTextSplitter
```

```python
text = pages[0].page_content
```

```python
#len은 어떤 기준으로 chunk size를 잴 것인가?의 기준이 되는 함수입니다.
#chunk_overlap은 chunk의 앞뒤로 다른 chunk와 설정한 size까지 겹칠 수 있도록 설정하는 것입니다.
text_splitter = CharacterTextSplitter(separator="\n\n", chunk_size=1000, chunk_overlap=100, length_function = len,)
chunks = text_splitter.split_text(text)
```

```python
print(chunks[0])
```

```python
length = []
for chunk in chunks:
    length.append(len(chunk))

print(length)
```

토큰 단위로 텍스트 분할해보기

```python
import tiktoken
tokenizer = tiktoken.get_encoding("cl100k_base")

def tiktoken_len(text):
    tokens = tokenizer.encode(
        text
    )
    return len(tokens)
```

```python
tiktoken_length = []
for chunk in chunks:
    tiktoken_length.append(tiktoken_len(chunk))

print(length)
print(tiktoken_length)
```

```python
import numpy as np
from numpy import dot
from numpy.linalg import norm
def cos_sim(A, B):
  return dot(A, B)/(norm(A)*norm(B))
```

Step 5 : Retriever 사용해보기

```python
from langchain_classic.chains.retrieval_qa.base import RetrievalQA
```

```python
from langchain_core.callbacks.streaming_stdout import StreamingStdOutCallbackHandler

# OpenAI 모델로 변경
# streaming=True와 callbacks 설정을 통해 실시간 출력을 활성화합니다.
llm = ChatOpenAI(
    model="gpt-4o",              # 또는 "gpt-4o-mini"
    temperature=0.0,
    streaming=True,              # 실시간 출력을 켭니다
    callbacks=[StreamingStdOutCallbackHandler()], # 출력을 콘솔에 바로 뿌려줍니다
)
```

```python
qa = RetrievalQA.from_chain_type(llm, chain_type="stuff",
                                 retriever=db.as_retriever(
                                     search_type="mmr",
                                     search_kwargs={"k": 3, "fetch_k" : 10}),
                                 return_source_documents=True)
```

```python
query = "예외를 발생시키지 않고 깨지는 비정형 문서 데이터를 어떤 지표와 기록 구조로 검증해야 검색·학습 품질을 보장할 수 있는가?"
result = qa.invoke({"query": query})   # 체인 실행은 invoke(). 입력은 {"query": 질문} 딕셔너리, 결과도 딕셔너리로 돌아온다
```

```python
RAG용 테스트셋 생성
전체 흐름
1. PDF를 LangChain Document로 로드합니다.
2. RAGAS가 문서를 쪼개 지식 그래프를 만듭니다. 요약, 헤드라인, 키워드, 엔티티를 뽑고 청크 사이의 관계를 연결합니다.
3. 그래프를 바탕으로 질문 유형별 시나리오를 만듭니다.
4. 시나리오마다 **질문(user_input), 정답(reference), 근거 컨텍스트(reference_contexts)**를 생성합니다.
```

2. PDF 로드

```python
from langchain_community.document_loaders import PyPDFLoader, DirectoryLoader

# PDF 하나
docs = PyPDFLoader("/content/drive/MyDrive/my-project/D41 Data Engineering.pdf").load()

# 폴더 전체
# docs = DirectoryLoader("./pdfs", glob="**/*.pdf", loader_cls=PyPDFLoader).load()

# 같은 파일에서 나온 페이지끼리 묶이도록 filename 메타데이터 지정
for d in docs:
    d.metadata["filename"] = d.metadata["source"]

# 빈 페이지나 너무 짧은 페이지 제거 (짧은 문서는 변환 단계에서 에러가 나기 쉬움)
docs = [d for d in docs if len(d.page_content.strip()) > 200]
```

3. LLM과 임베딩 설정

```python
from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from ragas.llms import LangchainLLMWrapper
from ragas.embeddings import LangchainEmbeddingsWrapper

generator_llm = LangchainLLMWrapper(ChatOpenAI(model="gpt-4o", temperature=0))
generator_emb = LangchainEmbeddingsWrapper(OpenAIEmbeddings(model="text-embedding-3-small"))
```

4. 질문 유형 비율 지정

```python
from ragas.testset.synthesizers import (
    SingleHopSpecificQuerySynthesizer,   # 한 청크로 답할 수 있는 구체적 질문
    MultiHopAbstractQuerySynthesizer,    # 여러 청크를 종합해야 하는 추상적 질문
    MultiHopSpecificQuerySynthesizer,    # 여러 청크를 연결해야 하는 구체적 질문
)

query_distribution = [
    (SingleHopSpecificQuerySynthesizer(llm=generator_llm), 0.5),
    (MultiHopAbstractQuerySynthesizer(llm=generator_llm), 0.25),
    (MultiHopSpecificQuerySynthesizer(llm=generator_llm), 0.25),
]
```

5. 한국어로 질문 생성하기

```python
for synth, _ in query_distribution:
    prompts = await synth.adapt_prompts("korean", llm=generator_llm)
    synth.set_prompts(**prompts)
```

6. 테스트셋 생성

```python
from ragas.testset import TestsetGenerator
#from ragas.testset.transforms import (
#    SummaryExtractor, NERExtractor, ThemesExtractor, EmbeddingExtractor,
#    CosineSimilarityBuilder, OverlapScoreBuilder, Parallel,
#)

from ragas.testset.transforms.extractors.llm_based import (
    SummaryExtractor, NERExtractor, ThemesExtractor,
)
from ragas.testset.transforms.extractors import EmbeddingExtractor
from ragas.testset.transforms.relationship_builders import (
    CosineSimilarityBuilder, OverlapScoreBuilder,
)
from ragas.testset.transforms.engine import Parallel


summary_extractor = SummaryExtractor(llm=generator_llm)
ner_extractor     = NERExtractor(llm=generator_llm)       # 엔티티 → SingleHop/MultiHopSpecific에 필요
themes_extractor  = ThemesExtractor(llm=generator_llm)    # 주제 → MultiHopAbstract에 필요
summary_emb = EmbeddingExtractor(
    embedding_model=generator_emb,
    property_name="summary_embedding",
    embed_property_name="summary",
)
cosine_sim  = CosineSimilarityBuilder(
    property_name="summary_embedding",
    new_property_name="summary_similarity",
    threshold=0.7,
)
ner_overlap = OverlapScoreBuilder(threshold=0.01)

transforms = [
    summary_extractor,
    Parallel(summary_emb, themes_extractor, ner_extractor),
    Parallel(cosine_sim, ner_overlap),
]

generator = TestsetGenerator(llm=generator_llm, embedding_model=generator_emb)
testset = generator.generate_with_langchain_docs(
    docs,
    testset_size=20,
    transforms=transforms,
    query_distribution=query_distribution,
)
```

```python
df = testset.to_pandas()
df.to_csv("testset.csv", index=False, encoding="utf-8-sig")  # 엑셀에서 한글이 깨지지 않도록
df[["user_input", "reference", "synthesizer_name"]].head()
```
